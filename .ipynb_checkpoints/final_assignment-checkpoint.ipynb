````python
# Fill in your selected classes
SELECTED_CLASSES = [
    'pad_thai',
    'spring_rolls',
    'fried_rice',
    'chicken_curry',
    'dumplings',
    'pulled_pork_sandwich',
    'caesar_salad',
    'gyoza',
    'peking_duck',
    'waffles',
    'chicken_wings',
    'steak'
]

invalid = [c for c in SELECTED_CLASSES if c not in CLASS_NAMES]
if invalid:
    print(f'WARNING: not in Food-101: {invalid}')
else:
    print(f'{len(SELECTED_CLASSES)} classes selected: all valid.')
````

## 1.3: Load a Small Subset for Fast Iteration

To figure out your model size you don't want to immediately train on all images in a class. Choose `N_PER_CLASS` images to quickly train a small model so you can optimize the model and the hyperparameters. Too big and you training will take long, too small and training might not work at all.


````python
N_PER_CLASS = 200  # images per class for fast iteration. ADJUST

transforms_pipeline = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

def filter_by_classes(dataset, selected):
    """Keep only selected classes and remap labels to 0..N-1."""
    invalid = [c for c in selected if c not in dataset.classes]
    if invalid:
        raise ValueError(f'Not in Food-101: {invalid}')
    selected_idx    = [dataset.classes.index(c) for c in selected]
    idx_map         = {old: new for new, old in enumerate(selected_idx)}
    mask            = [i for i, l in enumerate(dataset._labels) if l in selected_idx]
    dataset._labels = [idx_map[l] if l in idx_map else l for l in dataset._labels]
    return Subset(dataset, mask)

def subsample(dataset, n_per_class, num_classes):
    """Take at most n_per_class images per class."""
    counts  = {c: 0 for c in range(num_classes)}
    indices = []
    for i in dataset.indices:
        label = dataset.dataset._labels[i]
        if counts[label] < n_per_class:
            indices.append(i)
            counts[label] += 1
    return Subset(dataset.dataset, indices)

_train_full = datasets.Food101(root=DATA_DIR, split='train',
                                transform=transforms_pipeline, download=False)
_val_full   = datasets.Food101(root=DATA_DIR, split='test',
                                transform=transforms_pipeline, download=False)

train_filtered = filter_by_classes(_train_full, SELECTED_CLASSES)
val_filtered   = filter_by_classes(_val_full,   SELECTED_CLASSES)
NUM_CLASSES    = len(SELECTED_CLASSES)
CLASS_NAMES    = SELECTED_CLASSES

train_small = subsample(train_filtered, N_PER_CLASS, NUM_CLASSES)
val_small   = subsample(val_filtered,   N_PER_CLASS, NUM_CLASSES)

pin = DEVICE.type == 'cuda'
train_loader = DataLoader(train_small, batch_size=BATCH_SIZE,
                          shuffle=True,  num_workers=4, pin_memory=pin)
val_loader   = DataLoader(val_small,   batch_size=BATCH_SIZE,
                          shuffle=False, num_workers=4, pin_memory=pin)

print(f'Small subset - Train: {len(train_small):,} | Val: {len(val_small):,} | Classes: {NUM_CLASSES}')
````

## 1.4: Design Your CNN

Design and implement a CNN from scratch. Do not use pretrained weights.

Think about: how many convolutional layers do you need? How many filters? What pooling strategy? How do you move from convolutional layers to a classification head?

````python
class FoodNet(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2, 2), 

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.MaxPool2d(2, 2), 
            
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  

            nn.Conv2d(128, 256, kernel_size=3, padding=1),  
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),   
            nn.Dropout2d(p=0.25),

            nn.Conv2d(256, 512, kernel_size=3, padding=1),
            nn.BatchNorm2d(512),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),        
            nn.Dropout2d(p=0.3),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(512 * 4 * 4, 512),
            nn.ReLU(),
            nn.Dropout(p=0.5),
            nn.Linear(512, num_classes)
        )
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, nonlinearity='relu')
                nn.init.zeros_(m.bias)
            elif isinstance(m, nn.Linear):
                nn.init.kaiming_normal_(m.weight, nonlinearity='relu')
                nn.init.zeros_(m.bias)

    def forward(self, x):
        return self.classifier(self.features(x))

model = FoodNet(num_classes=NUM_CLASSES)
````

### Your Turn
Define your CNN and explain your choices:

I based the model off of the one described in Lu(2016)[1] whose premise is extremely similar to that of this assignment. However as the paper was published back in 2016 and lacks many modern features/techniques, changes have been made in order to modernize the model. The paper utilizes 3 convolution block layers which feeds into a fully connected layer, of which I have kept mostly the same, however changes where made to the convolution blocks themselves in order to include batch normalization within them as they are a standard feature in most modern CNN architectures. I have also switched from using a slowly decreasing kernel size to a standard uniform 3x3 kernel size throughout the model as it is considered the modern standard for CNNs after it's effectiveness compared to alternative kernel sizes have been proven with VGGNet. This arguably makes the downsizing less aggressive and I could counterbalance it by increasing the stride or adding more convolution blocks however at this stage I decided to test out the model first before making any changes. Relu was used as it is mostly standard for CNNs and was also used within the paper. We're using He initialization as it is the standard choice for RE. The dropout was kept the same as in the paper. The amount of parameters is a bit higher than reasonable at 4.3 million parameters however it is generally better to overfit as we can regulate it back down later. 

[1] Lu, Y. (2016). "Food Image Recognition by Using Convolutional Neural Networks (CNNs)," CoRR, vol. abs/1612.00983. [Online]. Available: http://arxiv.org/abs/1612.00983

## 1.5: Train and Optimize

Train your model on the small subset. Look at the loss curves and iterate: adjust your architecture, learning rate, regularization, or number of epochs. **Show at least two iterations and explain what you changed and why.**

Always reinitialise your model before each new training run so you are starting from a clean slate.

````python
def train_one_epoch(model, loader, optimizer, criterion, device):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for images, labels in loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)
        preds = outputs.argmax(dim=1)
        correct += (preds == labels).sum().item()
        total += labels.size(0)
    return running_loss / total, correct / total


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * images.size(0)
            preds = outputs.argmax(dim=1)
            correct += (preds == labels).sum().item()
            total += labels.size(0)
    return running_loss / total, correct / total


def plot_history(history, title='Training History'):
    epochs = range(1, len(history['train_loss']) + 1)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
    ax1.plot(epochs, history['train_loss'], label='Train')
    ax1.plot(epochs, history['val_loss'],   label='Val')
    ax1.set_title('Loss')
    ax1.set_xlabel('Epoch')
    ax1.legend()
    ax2.plot(epochs, history['train_acc'], label='Train')
    ax2.plot(epochs, history['val_acc'],   label='Val')
    ax2.set_title('Accuracy')
    ax2.set_xlabel('Epoch')
    ax2.legend()
    plt.suptitle(title)
    plt.tight_layout()
    plt.show()


def run_training(model, train_loader, val_loader, optimizer, criterion,
                 epochs, device, scheduler=None):
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    for epoch in range(epochs):
        train_loss, train_acc = train_one_epoch(model, train_loader, optimizer, criterion, device)
        val_loss,   val_acc   = evaluate(model, val_loader, criterion, device)
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        if scheduler:
            scheduler.step()
        print(f'Epoch {epoch+1:>3}/{epochs} | '
              f'Train Loss: {train_loss:.4f} Acc: {train_acc:.4f} | '
              f'Val Loss: {val_loss:.4f} Acc: {val_acc:.4f}')
    return history
````

````python
model     = FoodNet(num_classes=NUM_CLASSES).to(DEVICE)
criterion = nn.CrossEntropyLoss()
optimizer = optimizer = optim.AdamW(model.parameters(), lr=0.0001, weight_decay=0.01)
# optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9, weight_decay=0.0001, nesterov=True)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=70)

history = run_training(
    model, train_loader, val_loader,
    optimizer, criterion,
    epochs=70,
    device=DEVICE,
    scheduler=scheduler
)

plot_history(history, title='Iteration 1')

````
